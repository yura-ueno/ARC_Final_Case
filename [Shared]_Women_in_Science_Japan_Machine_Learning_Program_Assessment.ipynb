```python
import torch
from torch import nn, optim
from torch.nn import functional as F
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import v2
from tqdm.notebook import tqdm

config = {'image_size': 64, 'hidden_size': 32, 'epochs': 10, 'learning_rate': 1e-4, 'batch_size': 64}
torch.manual_seed(123)
```

```python
# Code provided to applicants
train_transforms = v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)])
test_transforms = v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)])
train_data = datasets.DTD('.', split='train', transform=train_transforms, download=True)
test_data = datasets.DTD('.', split='test', transform=test_transforms, download=True)

train_loader = DataLoader(train_data, batch_size=config['batch_size'], shuffle=True, drop_last=True, num_workers=2)
test_loader = DataLoader(test_data, batch_size=config['batch_size'], num_workers=2)
```

-  Add a third convolutional layer called `self.conv3` with the correct number of input and output channels, using the same kernel size and padding as the previous layers.
- Add a final linear layer called `self.fc2` with the correct input and output size.
- Initialize an Adam optimizer for the network using a learning rate equal to `config['learning_rate']`.

```python
# Code provided to applicants
class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(3, config['hidden_size'], kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(config['hidden_size'])
        self.conv2 = nn.Conv2d(config['hidden_size'], 2 * config['hidden_size'], kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(2 * config['hidden_size'])
        ### YOUR CODE HERE ###
        self.bn3 = nn.BatchNorm2d(4 * config['hidden_size'])
        self.fc1 = nn.Linear(4 * config['hidden_size'] * (config['image_size'] // 2 ** 3) ** 2, 8 * config['hidden_size'])
        ### YOUR CODE HERE ###

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = F.relu(F.max_pool2d(self.bn1(self.conv1(x)), kernel_size=2))
        x = F.relu(F.max_pool2d(self.bn2(self.conv2(x)), kernel_size=2))
        x = F.relu(F.max_pool2d(self.bn3(self.conv3(x)), kernel_size=2))
        x = torch.flatten(x, 1)  # Flatten (except for batch dimension)
        x = F.relu(F.dropout(self.fc1(x), p=0.5))
        x = self.fc2(x)
        return x

net = Net()
# Optimizer goes here
```

Using the provided `train` and `test` functions, train the network for `config['epochs']` epochs, and test and print the accuracy on the test set after each epoch.

```python
def train(net: Net, optimizer: optim.Optimizer, dataloader: DataLoader) -> float:
    net.train()
    losses = []
    for i, (inputs, labels) in tqdm(enumerate(dataloader), total=len(dataloader)):
        optimizer.zero_grad()
        outputs = net(inputs)
        loss = F.cross_entropy(outputs, labels)
        loss.backward()
        losses.append(loss.item())
        optimizer.step()
    return sum(losses) / len(losses)

def test(net: Net, dataloader: DataLoader) -> float:
    net.eval()
    correct = 0
    for inputs, labels in tqdm(dataloader, total=len(dataloader)):
        with torch.inference_mode():
            outputs = net(inputs)
        predicted = torch.argmax(outputs, 1)
        correct += (predicted == labels).sum().item()
    return correct / len(dataloader.dataset)
```
